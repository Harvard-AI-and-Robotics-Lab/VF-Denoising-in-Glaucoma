```python
import os
import numpy as np

data_path = os.path.expandvars("$VF_DENOISE_ROOT/dataset_train_val_test/N2N/fine_tune_dataset_Nge1.npz")

# Reproducibility
np.random.seed(42)

data = np.load(data_path)
keys = data.files
print("Keys:", keys)

# Number of samples (assume all arrays share the same first dim)
N = data[keys[0]].shape[0]
print("Total samples:", N)

# Shuffle indices
indices = np.random.permutation(N)

# 4:1 split
split = int(0.8 * N)
train_idx = indices[:split]
val_idx   = indices[split:]

# Split data
train_data = {k: data[k][train_idx] for k in keys}
val_data   = {k: data[k][val_idx] for k in keys}

# Save
np.savez("fine_tune_dataset_Nge1_train.npz", **train_data)
np.savez("fine_tune_dataset_Nge1_val.npz", **val_data)

print("Saved:")
print(f"  Train: {len(train_idx)} samples")
print(f"  Val:   {len(val_idx)} samples")
```
